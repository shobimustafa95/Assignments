```python
if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "x":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Cannot divide by zero.")
else:
    print("Invalid operator.")
```

# Qno:13

```python
salary = float(input("Enter salary: Rs. "))

if salary > 2500000:
    tax_rate = 0.15
elif salary >= 1500000:
    tax_rate = 0.10
elif salary >= 500000:
    tax_rate = 0.05
else:
    tax_rate = 0

tax = salary * tax_rate
gross_salary = salary - tax

print("Tax Deduction:", tax)
print("Gross Salary:", gross_salary)
```
